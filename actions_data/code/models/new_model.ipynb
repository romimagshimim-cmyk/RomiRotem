```python
# Full working example. Assumes `df` exists as your original pandas DataFrame.
# Adjust variable names if your DataFrame columns differ.

# -----------------------
# 0. Configuration
# -----------------------
# Columns that are counts (continuous, can be >1)
COUNT_COLS = ['foulCount', 'techFoulCount', 'netAlgaeCount', 'wallAlgaeCount']

# Embedding size for team id
EMBED_DIM = 32
SHARED_DIM = 128
DENSE_UNITS = [128, 256, 128]
EPOCHS = 20
BATCH_SIZE = 32

# -----------------------
# 1. Prepare dataframe
# -----------------------
# Copy and coerce numeric values; ensure team_number and position exist
df_clean = df.copy()
df_clean = df_clean.apply(pd.to_numeric, errors='coerce').astype('float32')
df_clean.fillna(0, inplace=True)

# Ensure team_number exists and is int
if 'team_number' not in df.columns:
    raise KeyError("DataFrame must include 'team_number' column.")
df_clean['team_number'] = df['team_number'].astype('int32')

# Create list of target columns (exclude identifiers)
exclude = ['team_number', 'possition', 'position']  # include any id cols you used
target_columns = [c for c in df_clean.columns if c not in exclude]

# Validate COUNT_COLS are present
for c in COUNT_COLS:
    if c not in target_columns:
        raise KeyError(f"Count column '{c}' not found in data columns.")

# Create binary_like and ordered target list (binary first, then counts)
binary_like = [c for c in target_columns if c not in COUNT_COLS]
ordered_target_columns = binary_like + COUNT_COLS

binary_dim = len(binary_like)
count_dim = len(COUNT_COLS)

# -----------------------
# 2. Prepare arrays for 3-team rows
#    Assumes rows are arranged in triples (team1, team2, team3) per alliance
# -----------------------
team1_idx = np.arange(0, len(df_clean), 3)
team2_idx = np.arange(1, len(df_clean), 3)
team3_idx = np.arange(2, len(df_clean), 3)

team1_ids = df_clean.iloc[team1_idx]['team_number'].values.reshape(-1, 1).astype('int32')
team2_ids = df_clean.iloc[team2_idx]['team_number'].values.reshape(-1, 1).astype('int32')
team3_ids = df_clean.iloc[team3_idx]['team_number'].values.reshape(-1, 1).astype('int32')

# Labels split by binary/count and ordered to match model outputs
team1_binary = df_clean.iloc[team1_idx][binary_like].values.astype('float32')
team1_count  = df_clean.iloc[team1_idx][COUNT_COLS].values.astype('float32')

team2_binary = df_clean.iloc[team2_idx][binary_like].values.astype('float32')
team2_count  = df_clean.iloc[team2_idx][COUNT_COLS].values.astype('float32')

team3_binary = df_clean.iloc[team3_idx][binary_like].values.astype('float32')
team3_count  = df_clean.iloc[team3_idx][COUNT_COLS].values.astype('float32')

# -----------------------
# 3. Build model
# -----------------------
# Inputs (unique names)
in_1 = Input(shape=(1,), name='team1_in')
in_2 = Input(shape=(1,), name='team2_in')
in_3 = Input(shape=(1,), name='team3_in')

# Embedding (shared)
max_team_id = int(df_clean['team_number'].max()) + 1
team_embedding = Embedding(input_dim=max_team_id, output_dim=EMBED_DIM, name='team_embedding')

e1 = Flatten(name='embed_flat_1')(team_embedding(in_1))
e2 = Flatten(name='embed_flat_2')(team_embedding(in_2))
e3 = Flatten(name='embed_flat_3')(team_embedding(in_3))

# Shared context
concat_all = Concatenate(name='alliance_concat')([e1, e2, e3])
shared = Dense(SHARED_DIM, activation='relu', name='shared_context')(concat_all)

# Team block factory (returns binary_out, count_out)
def team_block(team_embed, shared_context, prefix):
    x = Concatenate(name=f'{prefix}_concat')([team_embed, shared_context])
    x = Dense(DENSE_UNITS[0], activation='relu', name=f'{prefix}_dense1')(x)
    x = Dense(DENSE_UNITS[1], activation='relu', name=f'{prefix}_dense2')(x)
    x = Dense(DENSE_UNITS[2], activation='relu', name=f'{prefix}_dense3')(x)

    binary_out = Dense(binary_dim, activation='sigmoid', name=f'{prefix}_binary')(x)
    # Use linear activation (None) for counts so output range is unbounded; relu if you want non-negative
    count_out  = Dense(count_dim, activation='linear', name=f'{prefix}_count')(x)
    return binary_out, count_out

b1, c1 = team_block(e1, shared, 'team1')
b2, c2 = team_block(e2, shared, 'team2')
b3, c3 = team_block(e3, shared, 'team3')

model = Model(
    inputs={'team1_in': in_1, 'team2_in': in_2, 'team3_in': in_3},
    outputs={
        'team1_binary': b1, 'team1_count': c1,
        'team2_binary': b2, 'team2_count': c2,
        'team3_binary': b3, 'team3_count': c3
    },
    name='alliance_model'
)

# -----------------------
# 4. Compile model with appropriate losses
# -----------------------
model.compile(
    optimizer='adam',
    loss={
        'team1_binary': 'binary_crossentropy',
        'team2_binary': 'binary_crossentropy',
        'team3_binary': 'binary_crossentropy',
        'team1_count': 'mse',
        'team2_count': 'mse',
        'team3_count': 'mse'
    },
    loss_weights={
        'team1_binary': 1.0, 'team1_count': 1.0,
        'team2_binary': 1.0, 'team2_count': 1.0,
        'team3_binary': 1.0, 'team3_count': 1.0
    }
)

# Optional: inspect model structure
print(model.summary())

# -----------------------
# 5. Train model
# -----------------------
model.fit(
    x={'team1_in': team1_ids, 'team2_in': team2_ids, 'team3_in': team3_ids},
    y={
        'team1_binary': team1_binary, 'team1_count': team1_count,
        'team2_binary': team2_binary, 'team2_count': team2_count,
        'team3_binary': team3_binary, 'team3_count': team3_count
    },
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    validation_split=0.1,
    verbose=2
)

# -----------------------
# 6. Predict on a new alliance example and display nicely
# -----------------------
def display_team_predictions(team_preds, team_names, ordered_columns, per_line=5):
    for name, pred in zip(team_names, team_preds):
        print(f"\nTeam {name} predictions:")
        for i in range(0, len(ordered_columns), per_line):
            labels = []
            for j in range(i, min(i+per_line, len(ordered_columns))):
                col = ordered_columns[j]
                val = pred[j]
                labels.append(f"{col}: {val:.3f}")
            print("  " + " | ".join(labels))

# Example test alliance (use actual team ids from your dataset)
test_input = {
    'team1_in': np.array([[1323]], dtype='int32'),
    'team2_in': np.array([[254]], dtype='int32'),
    'team3_in': np.array([[1678]], dtype='int32')
}

preds = model.predict(test_input)

# Combine binary and count outputs in the same order as ordered_target_columns
team1_pred = np.concatenate([preds['team1_binary'][0], preds['team1_count'][0]])
team2_pred = np.concatenate([preds['team2_binary'][0], preds['team2_count'][0]])
team3_pred = np.concatenate([preds['team3_binary'][0], preds['team3_count'][0]])

display_team_predictions(
    team_preds=[team1_pred, team2_pred, team3_pred],
    team_names=['1323', '254', '1678'],
    ordered_columns=ordered_target_columns,
    per_line=6
)
```
