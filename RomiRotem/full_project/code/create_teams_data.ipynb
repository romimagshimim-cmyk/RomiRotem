```python
import requests
import pandas as pd
```

```python
AUTH_KEY = "Tnuge6o2FKoppKZ7DZaeFgbnyXRFivvrI6FPwao5XhhcPzkCOT2K525UTbqKr13o" 

headers = {
    "X-TBA-Auth-Key": AUTH_KEY.strip()
}

team_data = []
page_num = 0

while True:
    url = f"https://www.thebluealliance.com/api/v3/teams/{page_num}/simple"
    response = requests.get(url, headers=headers)
    if response.status_code != 200:
        break
        
    teams = response.json()
    if not teams:
        break

    for team in teams:
        team_data.append({
            "TeamNum": team.get("team_number"),
            "TeamName": team.get("nickname"),
            "Country": team.get("country")
        })
        
    page_num += 1

df = pd.DataFrame(team_data)

file_path = r"C:\Projects\Statbotics\RomiRotem\data\datasets\X.csv"
df.to_csv(file_path, index=False)
```
